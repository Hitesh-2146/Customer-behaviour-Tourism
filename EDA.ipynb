```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Set style for plots
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
```

```python
# ==========================================
# 1. LOAD DATA & INITIAL INSPECTION
# ==========================================
print("--- 1. LOADING DATA ---")
df = pd.read_csv("Customer behaviour Tourism.csv")

print("\nShape of Dataset:", df.shape)
print("\nFirst 5 rows:")
print(df.head())

print("\nDataset Info:")
print(df.info())

print("\nMissing Values Count:")
print(df.isnull().sum())

print("\nSummary Statistics (Numeric Columns):")
print(df.describe())
```

```python
# ==========================================
# 2. DATA CLEANING & PREPROCESSING
# ==========================================
print("\n--- 2. DATA CLEANING ---")
# Fill missing numerical values with median or zero, and categorical with mode or placeholder
# Here we fill numeric features with their median and categorical features with 'Unknown'
for col in df.select_dtypes(include=["float64", "int64"]).columns:
  df[col] = df[col].fillna(df[col].median())

for col in df.select_dtypes(include=["object"]).columns:
  df[col] = df[col].fillna("Unknown")

print("Missing values after cleaning:", df.isnull().sum().sum())
```

```python
# ==========================================
# 3. UNIVARIATE ANALYSIS
# ==========================================
print("\n--- 3. UNIVARIATE ANALYSIS ---")

# Target Variable Distribution
plt.figure(figsize=(6, 4))
sns.countplot(
    data=df, x="Taken_product", palette="Set2", hue="Taken_product", legend=False
)
plt.title("Distribution of Target Variable (Taken_product)")
plt.savefig("target_distribution.png", bbox_inches="tight")
plt.close()

# Histograms for numerical columns
num_cols = [
    "Yearly_avg_view_on_travel_page",
    "total_likes_on_outstation_checkin_given",
    "Yearly_avg_comment_on_travel_page",
    "Daily_Avg_mins_spend_on_traveling_page",
]
df[num_cols].hist(bins=20, figsize=(12, 10), color="skyblue", edgecolor="black")
plt.suptitle("Histograms of Key Numeric Features", fontsize=16)
plt.savefig("numeric_histograms.png", bbox_inches="tight")
plt.close()

```

```python

# ==========================================
# 4. BIVARIATE ANALYSIS (FEATURE vs TARGET)
# ==========================================
print("\n--- 4. BIVARIATE ANALYSIS ---")

# Numerical feature vs Target Boxplots
plt.figure(figsize=(10, 6))
sns.boxplot(
    data=df,
    x="Taken_product",
    y="Daily_Avg_mins_spend_on_traveling_page",
    palette="Pastel1",
    hue="Taken_product",
    legend=False,
)
plt.title("Daily Avg Mins Spent vs Product Purchase")
plt.savefig("mins_vs_product.png", bbox_inches="tight")
plt.close()

# Categorical feature vs Target (e.g., Preferred Device)
plt.figure(figsize=(10, 6))
sns.countplot(
    data=df, x="preferred_device", hue="Taken_product", palette="Set1"
)
plt.title("Preferred Device vs Product Purchase")
plt.xlabel("Preferred Device")
plt.ylabel("Count")
plt.legend(title="Taken Product")
plt.savefig("device_vs_product.png", bbox_inches="tight")
plt.close()

```

```python

# ==========================================
# 5. CORRELATION ANALYSIS
# ==========================================
print("\n--- 5. CORRELATION ANALYSIS ---")

# Encode target variable for correlation
df_numeric = df.select_dtypes(include=[np.number]).copy()
if "Taken_product" in df.columns:
  df_numeric["Taken_product_num"] = df["Taken_product"].apply(
      lambda x: 1 if x == "Yes" else 0
  )

plt.figure(figsize=(12, 8))
corr_matrix = df_numeric.corr()
sns.heatmap(
    corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", linewidths=0.5
)
plt.title("Correlation Matrix Heatmap")
plt.savefig("correlation_heatmap.png", bbox_inches="tight")
plt.close()

print("\nEDA Completed Successfully! All plots saved as PNG images.")
```
